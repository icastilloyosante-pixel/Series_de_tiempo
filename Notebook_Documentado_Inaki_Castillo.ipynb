# 1. Caminata Aleatoria (Random Walk)

Una caminata aleatoria se define como:

$$
X_t = X_{t-1} + Z_t
$$

donde:

- $X_t$ es el valor de la serie en el tiempo $t$.
- $Z_t$ es un ruido blanco:

$$
Z_t \sim N(0,\sigma^2)
$$

## Propiedades

- No estacionaria en niveles.
- La varianza aumenta con el tiempo.
- Es la base de la Hipótesis de Mercado Eficiente.

La primera diferencia transforma la serie en:

$$
\Delta X_t = X_t - X_{t-1}
$$

que corresponde al ruido blanco original.



# 2. Series Financieras

En este ejercicio se descargan precios históricos de Amazon y se estudian sus retornos.

## Retornos Logarítmicos

Los retornos logarítmicos se calculan mediante:

$$
r_t=\ln\left(\frac{P_t}{P_{t-1}}\right)
$$

donde:

- $P_t$ = precio en el periodo actual.
- $P_{t-1}$ = precio en el periodo anterior.

### Ventajas

1. Son aditivos en el tiempo.
2. Facilitan el modelado estadístico.
3. Se utilizan ampliamente en finanzas cuantitativas.



# 3. Estacionalidad y Variables Exógenas

Una serie temporal puede descomponerse en:

$$
Y_t = T_t + S_t + E_t
$$

donde:

- $T_t$ = tendencia.
- $S_t$ = componente estacional.
- $E_t$ = componente aleatoria.

En el notebook se agrega una variable exógena de promociones.

El modelo SARIMAX permite incorporar dichas variables:

$$
Y_t = SARIMA + \beta X_t
$$

donde $X_t$ representa variables externas relevantes.



# 4. Modelo SARIMA

SARIMA significa:

**Seasonal AutoRegressive Integrated Moving Average**

Su notación es:

$$
(p,d,q)(P,D,Q)_m
$$

donde:

### Componente no estacional

- $p$ = términos autoregresivos.
- $d$ = diferenciaciones.
- $q$ = términos de media móvil.

### Componente estacional

- $P$ = autoregresivo estacional.
- $D$ = diferenciación estacional.
- $Q$ = media móvil estacional.
- $m$ = periodicidad.

En el notebook original se utiliza:

$$
SARIMA(0,1,1)(0,1,1)_{12}
$$

con periodicidad mensual.



# 5. Pronósticos

Una vez estimado el modelo SARIMA se generan predicciones futuras:

$$
\hat{Y}_{t+h}
$$

para horizontes de pronóstico $h=1,2,\dots,n$.

La precisión del modelo depende de:

- Correcta especificación.
- Estacionariedad.
- Calidad de los datos.
- Ausencia de patrones en los residuos.



# 6. Análisis de Residuos

Un buen modelo debe generar residuos similares a ruido blanco.

Formalmente:

$$
e_t = Y_t - \hat{Y}_t
$$

Se espera que:

$$
E[e_t]=0
$$

y

$$
Cov(e_t,e_{t-k}) = 0
$$

para cualquier rezago $k$.

Las gráficas de residuos permiten validar si el modelo capturó adecuadamente la estructura de la serie.
